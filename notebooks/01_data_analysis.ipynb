```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/transactions.csv")

print("Dataset shape:", df.shape)

df.head()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
print(
    "Duplicate transaction IDs:",
    df["transaction_id"].duplicated().sum()
)
```

```python
print("Fraud distribution:")
print(df["fraud"].value_counts())

print("\nFraud percentage:")
print(
    df["fraud"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
fraud_comparison = df.groupby("fraud")[
    [
        "amount",
        "amount_deviation",
        "transactions_last_10min",
        "transactions_last_1h",
        "transactions_last_24h",
        "device_age_days",
        "location_distance_km",
        "ip_risk_score",
        "failed_attempts_24h",
        "previous_fraud_count",
        "account_velocity_score"
    ]
].mean().round(2)

fraud_comparison
```

```python
print("Average transaction amount:")
print(df.groupby("fraud")["amount"].mean().round(2))

print("\nAverage transactions in last 10 minutes:")
print(df.groupby("fraud")["transactions_last_10min"].mean().round(2))

print("\nAverage IP risk score:")
print(df.groupby("fraud")["ip_risk_score"].mean().round(3))
```
